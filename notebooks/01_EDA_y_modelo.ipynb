# EDA y modelo de riesgo de roya
Ejecute antes: `python -m cafe.data_download && python -m cafe.data_prep`.

```
import pandas as pd, plotly.express as px
from cafe import config as C
w = pd.read_csv(C.WEEKLY, parse_dates=['semana'])
w.describe(include='all').T
```

```
print(w.groupby('municipio').agg(semanas=('semana','count'), desde=('semana','min'), hasta=('semana','max'), lluvia=('lluvia_mm','mean'), temp=('temp_c','mean')).round(1))
```

```
px.line(w, x='semana', y='lluvia_mm', color='municipio', title='Lluvia semanal (mm)')
```

```
px.box(w.assign(mes=w.semana.dt.month), x='mes', y='lluvia_mm', title='Estacionalidad de la lluvia')
```

## Modelo vs. líneas base

```
from cafe import risk_model as rm
res = rm.evaluate(w)
pd.DataFrame(res['resultados'])
```

La variable no observada se marginaliza (eliminación de variables).
